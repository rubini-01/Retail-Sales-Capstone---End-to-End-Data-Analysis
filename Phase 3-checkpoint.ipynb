```python
import pandas as pd
```

```python
sales = pd.read_csv(r'C:\Users\ACER\Downloads\capstone project\source\retail_sales_2023.csv')
customers = pd.read_csv(r'C:\Users\ACER\Downloads\capstone project\source\customers.csv')
returns = pd.read_csv(r'C:\Users\ACER\Downloads\capstone project\source\returns.csv')
products = pd.read_csv(r'C:\Users\ACER\Downloads\capstone project\source\products.csv')
```

```python
sales_customers_df = pd.merge(
    sales,
    customers,
    on='Customer_ID',
    how='inner'
)
```

```python
sales_return_df = pd.merge(
    sales,
    returns,
    on='Order_ID',
    how='inner'
)
```

```python
## mean meridan and standard deviation of sales and profits
```

```python
print('Sales Statistics')
print('Mean:', sales['Sales'].mean())
print('Median:', sales['Sales'].median())
print('Std Dev:', sales['Sales'].std())

print('\nProfit Statistics')
print('Mean:', sales['Profit'].mean())
print('Median:', sales['Profit'].median())
print('Std Dev:', sales['Profit'].std())
```

```python
#IQR Outlier detection for sales
```

```python
Q1 = sales['Sales'].quantile(0.25)
Q3 = sales['Sales'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = sales[
    (sales['Sales'] < lower_bound) |
    (sales['Sales'] > upper_bound)
]

print('\nOutliers Detection for Sales:\n')
print(outliers.head(15))
```

```python
# Average sales and Order Frequencies
```

```python
Groupby_Segment = sales_customers_df.groupby('Segment').agg(
    Avg_Sales=('Sales', 'mean'),
    Order_Frequency=('Order_ID', 'count')
)

print('Groupby_Segment:\n')
print(Groupby_Segment)
```

```python
# Age Distribution
```

```python
age_groups = pd.cut(
    sales_customers_df['Age'],
    bins=[18, 25, 35, 45, 55, 65, 110],
    labels=['18-24', '25-34', '35-44', '45-54', '55-64', '65-110']
)

print('\nAge Group Distribution:\n')
print(age_groups.value_counts())
```

```python
# HeatMap Data
```

```python
city_heatmap_data = sales_customers_df.pivot_table(
    values='Sales',
    index='City',
    columns='Segment',
    aggfunc='mean'
).round(1)

print('\nCity-Level Heatmap Data:\n')
print(city_heatmap_data)
```

```python
# Line Chart : Monthly Sales Trend
```

```python
import datetime
```

```python
sales['Order_Date'] = pd.to_datetime(sales['Order_Date'])
```

```python
monthly_sales = sales.groupby(
    sales['Order_Date'].dt.month
)['Sales'].sum()

plt.figure(figsize=(8,5))

monthly_sales.plot(kind='line', marker='o')

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')

plt.show()
```
